## Bining (Discretization)
Discretization means taking continuous numbers (like exact ages, prices, or temperatures) and grouping them into fixed buckets or categories.
Continuous data (Exact numbers): Age = 12, 18, 25, 42, 67, 80

Discretized/Binned data (Categories):

Bucket 1 (0–17): "Child"

Bucket 2 (18–59): "Adult"

Bucket 3 (60+): "Senior"

Instead of tracking every single decimal or exact number, you just assign each value to its corresponding "bin."

# The Reason we use it
1. To Handle Outliers
Extreme, unusual numbers (outliers) can pull a model off track.

Example: If most people in a dataset earn $30,000 to $80,000, but one person earns $10,000,000, that $10,000,000 is an outlier.

How binning helps: If you group incomes into high-level categories like Low, Medium, and High, that $10,000,000 simply falls into the High bin along with $150,000. The extreme value loses its power to distort your model's calculations.

2. To Improve the Value Spread
Sometimes raw data is skewed, chaotic, or spread unevenly, making it hard for machine learning algorithms (especially linear models or tree-based algorithms) to find clean patterns.
