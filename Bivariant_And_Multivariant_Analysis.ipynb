```python
tips = sns.load_dataset('tips')   # tips is a built in data_set in seaborn
flight = sns.load_dataset('flights')
iris =  sns.load_dataset('iris')
titanic = pd.read_csv('train.csv')
```

## Scatterplot (numerical-numerical columns)

```python
sns.scatterplot(data=tips,x='total_bill', y = 'tip') # Bivariant analysis
```

# we can also draw a scatterplot among multiple columns (Multivariant)

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='smoker')
# hue is an additional peremiter in seaborn which allows you to analyse more than two columns
# style allows you to add 4th column i-e we add smoker here 
```

# Graph Explanation: #
In this scatterplot, the blue dots represent male customers and the orange dots represent female customers.
To dig deeper into the habits of these customers:
Circles (●) represent customers who are smokers.
Crosses (✖) represent customers who are non-smokers.
Key Observations:
Positive Correlation: There is a clear upward trend; as the total_bill increases, the tip amount generally increases as well.
Gender Distribution: While both genders are spread throughout the data, the highest bills and highest tips in this specific dataset appear to be paid by male customers (the blue points in the top right).

## Pair Plot ##
A pairplot in Seaborn is a powerful visualization tool used to understand the relationships between all numerical variables in a dataset at once. It creates a matrix of plots where each variable is shared across both the x and y axes.


```python
sns.pairplot(iris)
```

## Explaination of the PairPlot graphs # 
This is a pairplot of the Iris dataset. It acts like a "birds-eye view" of your entire data table.
Here is the simple breakdown:
The Grid: It compares every numeric column against every other numeric column.
The Scatter Plots: These show the relationship between two different features. For example, where petal_length and petal_width meet, you can see a strong upward trend, meaning as one gets bigger, the other does too.
The Histograms (The Diagonal): When a variable is compared with itself (down the middle), it shows a bar chart of its distribution—basically, how many flowers have certain measurements.
The Groups: You can see two distinct "clumps" of dots in many plots. This tells you there are likely different groups (species) within the data that have very different sizes.

It helps you quickly see which features are related and how the data is spread out

## Line Plot

```python
new
```

```python
sns.lineplot(data=new,x='year',y='passengers')
```
