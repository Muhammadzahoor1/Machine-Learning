# Feature Scaling #

Feature Scaling is a data preprocessing step used to bring all independent variables (features) down to a similar scale or size.
It has mainly Two types 
1) * Standardization 
2) * Normalization


# Standardization
Standardization is a feature scaling technique that transforms your data so that it has a Mean of 0 and a Standard Deviation of 1.

* Formula:
      _
  X - X divided by standard diviation
  where x= the original value,
  x_bar = Mean

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv('Social_Network_Ads.csv')
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
Scaling=StandardScaler()
```

```python
new= df[['Age', 'EstimatedSalary']]
new.head()
```

```python
scaled_data = Scaling.fit_transform(new[['Age','EstimatedSalary']])
```

```python
print("Mean: ", scaled_data.mean(axis=0))
```

# Normalization (Min-Max Normalization)
In this approach we will scale down the values of the deautures between 0 an 1

## Formula: 

        Normalization = X - X_mean Divided by X_max - X_min

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaling=MinMaxScaler()      # Creaing an object of minmaxScaler 
```

```python
scaling.fit_transform(new[['Age','EstimatedSalary']])
```

Converting the Scaled columns into DataFrame

```python
scaled_array = scaling.fit_transform(new[['Age', 'EstimatedSalary']])

scaled_dataFrame= pd.DataFrame(scaled_array,columns=['Age','EstimatedSalary'])
```

```python
np.round(scaled_dataFrame.describe())
```

```python
ax1 = new.plot(kind='scatter',x='Age',y='EstimatedSalary',color='red')
ax1.set_title("Before Scaling")

bx = scaled_dataFrame.plot(kind='scatter',x='Age',y='EstimatedSalary')
bx.set_title("After Scaling")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Set up a figure with 2 subplots side-by-side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# --- LEFT PLOT: Before Scaling ---
# Using the original 'new' DataFrame
sns.kdeplot(data=new, x='Age', ax=ax1, label='Age', fill=True, color='blue', alpha=0.4)
sns.kdeplot(data=new, x='EstimatedSalary', ax=ax1, label='Estimated Salary', fill=True, color='orange', alpha=0.4)
ax1.set_title('Before Scaling')
ax1.set_xlabel('Original Scale Value')
ax1.set_ylabel('Density')
ax1.legend()

# --- RIGHT PLOT: After Scaling ---
# Using the 'scaled_dataFrame' DataFrame
sns.kdeplot(data=scaled_dataFrame, x='Age', ax=ax2, label='Age (Scaled)', fill=True, color='blue', alpha=0.4)
sns.kdeplot(data=scaled_dataFrame, x='EstimatedSalary', ax=ax2, label='Estimated Salary (Scaled)', fill=True, color='orange', alpha=0.4)
ax2.set_title('After Scaling')
ax2.set_xlabel('Scaled Value')
ax2.set_ylabel('Density')
ax2.legend()

# Adjust layout to prevent overlap and display
plt.tight_layout()
plt.show()
```
