```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/fivethirtyeight/uber-tlc-foil-response/master/uber-trip-data/uber-raw-data-apr14.csv"
```

```python
df = pd.read_csv(url , nrows=2000)
```

```python
df.head()
```

```python
df.info()
```

```python
# coverting into datetime
df['Date/Time'] = pd.to_datetime(df['Date/Time'])
```

# Extracting Time

```python
df['time'] = df['Date/Time'].dt.time
df.sample(5)
```

# Extract Day

```python
df['day'] = df['Date/Time'].dt.day_name()
df.sample(5)
```

# Extract Month

```python
df['Month'] = df['Date/Time'].dt.month_name()
df.sample(5)
```

# Extract Year

```python
df['year'] = df['Date/Time'].dt.year
df.sample(20)
```

```python
df['Is_weakend'] = np.where(df['day'].isin(['Sunday' , 'Saturday']),1,0)
df.sample(5)
```

# Working With Time 

```python
df['Hours'] = df['Date/Time'].dt.hour
df['Minutes'] = df['Date/Time'].dt.minute
df['secands'] = df['Date/Time'].dt.second

df.sample(5)
```
